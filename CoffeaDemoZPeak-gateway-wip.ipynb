```python
# Make the CMS proxy + CA certs visible to the client and to ALL workers,
# including workers that join later (scale() starts them asynchronously via
# HTCondor). A WorkerPlugin's setup() runs on every worker at start, before
# XRootD is first used in the process; a plain client.run() only reaches the
# workers connected at that moment and silently misses the rest.
from distributed.diagnostics.plugin import WorkerPlugin

GRID_ENV = {
    "X509_USER_PROXY": "/home/jmustafi/.globus/x509up",
    "X509_CERT_DIR":   "/cvmfs/grid.cern.ch/etc/grid-security/certificates",
    "X509_VOMS_DIR":   "/cvmfs/grid.cern.ch/etc/grid-security/vomsdir",
}

class GridEnvPlugin(WorkerPlugin):
    name = "grid-env"
    def setup(self, worker):
        import os
        os.environ.update(GRID_ENV)

import os
os.environ.update(GRID_ENV)                      # this notebook (client)

try:
    client.register_plugin(GridEnvPlugin())      # distributed >= 2023.9
except AttributeError:
    client.register_worker_plugin(GridEnvPlugin())
print("grid-env plugin registered: current and future workers get the X509 env")
```

```python
import numpy as np
import awkward as ak
import matplotlib.pyplot as plt
%matplotlib inline
import hist
import coffea.processor as processor
from coffea.nanoevents import schemas
```

```python
class Processor(processor.ProcessorABC):
    def __init__(self):
        pass

    def process(self, events):
        dataset = events.metadata["dataset"]
        mu = events.Muon
        dimu_neutral = mu[(ak.num(mu) == 2) & (ak.sum(mu.charge, axis=1) == 0)]
        dimu_mass = (dimu_neutral[:, 0] + dimu_neutral[:, 1]).mass

        h = (
            hist.Hist.new
            .StrCat([], name="dataset", label="Dataset", growth=True)
            .Reg(50, 20, 120, name="dimu_mass", label=r"$\mu\mu$ Mass [GeV]")
            .Double()
        )
        h.fill(dataset=dataset, dimu_mass=dimu_mass)
        return {"dimu_mass": h}

    def postprocess(self, accumulator):
        pass
```
